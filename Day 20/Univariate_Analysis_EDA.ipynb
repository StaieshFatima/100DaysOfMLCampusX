### **Count Plot**

A count plot in Seaborn shows how many times each category/value occurs in a column.

How many people survived and how many didn't

```python
import matplotlib.pyplot as plt
```

```python
sns.countplot(x=df['Survived'])

plt.show()
```

```python
# To see how many people survived and how many died
df['Survived'].value_counts()
```

```python
df['Survived'].value_counts().plot(kind='bar')
plt.show()
```

Why plt.show()?

Seaborn creates the plot while Matplotlib displays it

```python
print(df['Pclass'].value_counts())
sns.countplot(x=df['Pclass'])
plt.show()
```

```python
sns.countplot(x=df['Sex'])
plt.show()
```

```python
sns.countplot(x=df['Embarked'])
plt.show()
```

**Pie Chart**

A pie chart is a circular chart divided into slices, each slice represents a percentage of the whole.

```python
df['Survived'].value_counts().plot(kind='pie',autopct="%.2f")
plt.show()
```

```python
df['Sex'].value_counts().plot(kind='pie',autopct="%.2f")
plt.show()
```
