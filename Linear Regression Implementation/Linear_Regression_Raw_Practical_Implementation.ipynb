```python
X=[1,2,3,4]
Y=[30,40,50,65]
```

### Calculating mean of X and Y

```python
X_mean=sum(X)/len(X)
X_mean
```

```python
Y_mean=sum(Y)/len(Y)
Y_mean
```

```python
X_der=[]
for x in X:
  X_der.append(x-X_mean)
print(f'X_derivation {X_der}')
```

```python
Y_der=[]
for y in Y:
  Y_der.append(y-Y_mean)
print(f'Y_derivation {Y_der}')
```

```python
product=X_der*Y_der
product
```

```python
result=[]
for x,y in zip(X_der,Y_der):
  result.append(x*y)
print(result)
```

```python
X_der_square=[]
for i in X_der:
  X_der_square.append(pow(i,2))
X_der_square
```

### **Slope Calculation**

```python
numerator=sum(result)
denominator=sum(X_der_square)

slope=numerator/denominator
slope
```

### **Intercept Calculation**

```python
mx_mean=slope*X_mean
intercept=Y_mean-mx_mean
intercept
```

### **Calculating Predicted Y**

y=mX+c

```python
y_pred=[]
for x in X:
  y_pred.append(slope*x+intercept)
print(f'predicted Y: {y_pred}')
```

To find residual error

error=y-y_pred

```python
error=[]

for actual,predicted in zip(Y,y_pred):
  error.append(actual-predicted)
print(f"Residual error {error}")
```

Calculating mean absolute error or MAE

```python
absolute=[]
for e in error:
  absolute.append(abs(e))
  MAE=sum(absolute)/len(absolute)
print("Mean Absolute error ",MAE)
```

Mean Square Error

```python
error_squared=[]
for a in error:
  error_squared.append(pow(a,2))
  MSE=sum(error_squared)/len(absolute)
print("Mean Square Error: ",MSE)
```

Root Mean Squared Error

```python
import math
RMSE=math.sqrt(MSE)
RMSE
```
