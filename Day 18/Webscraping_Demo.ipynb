There is this policy known as robot policy if I only request like from webpage it will think that I am acting as a bot ....therefore I need to send some header to make webpage think like server is requesting data

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/142.0.0.0 Safari/537.36'
}
requests.get("https://en.wikipedia.org/wiki/Hania_Aamir",headers=headers)
```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/142.0.0.0 Safari/537.36'
}
webpage=requests.get("https://en.wikipedia.org/wiki/Hania_Aamir",headers=headers).text
webpage
```

```python
soup=BeautifulSoup(webpage,'lxml')
```
